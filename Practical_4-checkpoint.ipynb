```python
#Import & Load Data

import pandas as pd

df = pd.read_csv(r"C:\Users\user\Desktop\Mall_Customers.csv")
print(df.head())
print(df.columns)
```

```python
#Create Data Subsets

subset_cols = df[['CustomerID','Age','Genre']]
subset_cols.head()
subset_rows = df[df['Annual Income (k$)'] < 100]
subset_rows.head()
```

```python
#Merge Data

df2 = df[['Genre', 'Annual Income (k$)']]

merged_df = pd.merge(df, df2, on='Genre')
merged_df.head()
```

```python
#Sort data

sorted_df = df.sort_values(by='Age', ascending=True)
sorted_df.head()
```

```python
#Transpose data 

transposed_df = df.head().T
transposed_df
```

```python
#Shape data

df.shape
```

```python
#Reshape data 

melted_df = pd.melt(df, id_vars=['CustomerID'], 
                    value_vars=['Age', 'Annual Income (k$)'])
melted_df.head()
```
