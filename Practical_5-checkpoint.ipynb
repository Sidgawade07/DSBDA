```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
df1 = pd.read_csv(r'C:\Users\user\Desktop\AirQuality.csv', sep=';')
df2 = pd.read_csv(r'C:\Users\user\Desktop\heart.csv')
```

```python
print(df1.columns)
df1.head()
```

```python
print(df2.columns)
df2.head()
```

```python
#data cleaning
print("Before Cleaning",df1.shape,df2.shape)
df1=df1.drop_duplicates()
df2=df2.drop_duplicates()
df1=df1.fillna(df1.mean(numeric_only=True))
df2=df2.fillna(df2.mean(numeric_only=True))
print("After Cleaning",df1.shape,df2.shape)
```

```python
#data Integration
df1['id']=range(len(df1))
df2['id']=range(len(df2))
data=pd.merge(df1,df2,on='id')
data.head()
```

```python
#data Transformation
data=data.select_dtypes(include='number')
data=data.fillna(0)
X=data.drop('target',axis=1)
y=data['target']
scaler=StandardScaler()
X=scaler.fit_transform(X)
```

```python
#Error Correction
data=data[(data>=0).all(axis=1)]
print("After Error Correction",data.shape)
```

```python
# data Model Building
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X = data.drop('target', axis=1)
y = data['target']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LogisticRegression(max_iter=2000, solver='liblinear')

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```
